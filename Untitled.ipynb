# Encoding  
# Binary Encoder or OHE(get_dummies) used for nominal 
# Label Encoder(sklearn)  used for ordinal values like 0, 1, 2 represent some order

```python
df.head()
df.columns
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

le = LabelEncoder()
df["Education_Level"] = le.fit_transform(df["Education_Level"])
df["Loan_Approved"] = le.fit_transform(df["Loan_Approved"])

```

```python
df.sample(5)
```

```python
## these are columns where we will perform OHE
cols = ["Employment_Status","Marital_Status","Loan_Purpose","Property_Area","Gender","Employer_Category"]

ohe = OneHotEncoder(drop = "first", sparse_output = False, handle_unknown = "ignore")
encoded = ohe.fit_transform(df[cols])
# this is in 2D array we will convert it into dataframe
encoded_df = pd.DataFrame(encoded, columns = ohe.get_feature_names_out(cols), index = df.index)

df = pd.concat([df.drop(columns=cols), encoded_df], axis = 1) ## removing the original columns as we have already done OHE
```

```python
df.head()
df.info() ## converted all object Dtype to numerical values
```

# Feature Scaling + Train test split 

```python
X = df.drop(columns = ["Loan_Approved"])
Y = df["Loan_Approved"]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.2, random_state = 42)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled
```

```python
X_test_scaled
```

# Training & Evaluation of models

```python
from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression()
lr_model.fit(X_train_scaled, Y_train)
Y_pred = lr_model.predict(X_test_scaled)
print("LOGISTIC REGRESSION MODEL :")
print("Accuracy Score: ", accuracy_score(Y_test, Y_pred))
print("Precision Score: ", precision_score(Y_test, Y_pred))
print("Recall Score: ", recall_score(Y_test, Y_pred))
print("F1 Score: ", f1_score(Y_test, Y_pred))
print("Confusion Matrix: ", confusion_matrix(Y_test, Y_pred))
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn_model = KNeighborsClassifier(n_neighbors = 5)
knn_model.fit(X_train_scaled, Y_train)
Y_pred = knn_model.predict(X_test_scaled)
print("KNN MODEL :")
print("Accuracy Score: ", accuracy_score(Y_test, Y_pred))
print("Precision Score: ", precision_score(Y_test, Y_pred))
print("Recall Score: ", recall_score(Y_test, Y_pred))
print("F1 Score: ", f1_score(Y_test, Y_pred))
print("Confusion Matrix: ", confusion_matrix(Y_test, Y_pred))
```

```python
from sklearn.naive_bayes import GaussianNB

nb_model = GaussianNB()
nb_model.fit(X_train_scaled, Y_train)
Y_pred = nb_model.predict(X_test_scaled)
print("KNN MODEL :")
print("Accuracy Score: ", accuracy_score(Y_test, Y_pred))
print("Precision Score: ", precision_score(Y_test, Y_pred))
print("Recall Score: ", recall_score(Y_test, Y_pred))
print("F1 Score: ", f1_score(Y_test, Y_pred))
print("Confusion Matrix: ", confusion_matrix(Y_test, Y_pred))
```

## Best Model on basis of precision => Naive Bayes

# Feature Engineering

```python
## credit_score and dti ratio were too much affecting as observed in coorelation heatmap

# if sometime data is skewed we do log of it
# x log = log(1+x) adding 1 to prevent log 0 condition
```

```python
#add or transform feature 
df["DTI_Ratio_sq"] = df["DTI_Ratio"] ** 2
df["Credit_Score_sq"] = df["Credit_Score"] ** 2

# df["Applicant_Income_log"] = np.log1p(df["Applicant_Income"])

X = df.drop(columns =["Loan_Approved", "Credit_Score", "DTI_Ratio"])
Y = df["Loan_Approved"]

X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.2, random_state = 42)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
lr_model = LogisticRegression()
lr_model.fit(X_train_scaled, Y_train)
Y_pred = lr_model.predict(X_test_scaled)
print("LOGISTIC REGRESSION MODEL :")
print("Accuracy Score: ", accuracy_score(Y_test, Y_pred))
print("Precision Score: ", precision_score(Y_test, Y_pred))
print("Recall Score: ", recall_score(Y_test, Y_pred))
print("F1 Score: ", f1_score(Y_test, Y_pred))
print("Confusion Matrix: ", confusion_matrix(Y_test, Y_pred))
```

```python
knn_model = KNeighborsClassifier(n_neighbors = 5)
knn_model.fit(X_train_scaled, Y_train)
Y_pred = knn_model.predict(X_test_scaled)
print("KNN MODEL :")
print("Accuracy Score: ", accuracy_score(Y_test, Y_pred))
print("Precision Score: ", precision_score(Y_test, Y_pred))
print("Recall Score: ", recall_score(Y_test, Y_pred))
print("F1 Score: ", f1_score(Y_test, Y_pred))
print("Confusion Matrix: ", confusion_matrix(Y_test, Y_pred))
```

```python
nb_model = GaussianNB()
nb_model.fit(X_train_scaled, Y_train)
Y_pred = nb_model.predict(X_test_scaled)
print("NAIVE BAYES MODEL :")
print("Accuracy Score: ", accuracy_score(Y_test, Y_pred))
print("Precision Score: ", precision_score(Y_test, Y_pred))
print("Recall Score: ", recall_score(Y_test, Y_pred))
print("F1 Score: ", f1_score(Y_test, Y_pred))
print("Confusion Matrix: ", confusion_matrix(Y_test, Y_pred))
```
